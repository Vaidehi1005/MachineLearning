```python
import pandas as pd
```

```python
delivery = pd.read_csv('deliveries.csv')
```

```python
delivery
```

```python
death_over = delivery[delivery['over']>15]
```

```python
name = death_over.groupby('batsman')['batsman_runs'].count()
x = name>200
name[x].shape
name[x]
```

```python
name[x].index
```

```python
name[x].values
```

```python
bool_list=name[x].index.tolist()
```

```python
delivery['batsman']
```

```python
delivery[delivery['batsman'].isin(bool_list)].shape
```

```python
final = delivery[delivery['batsman'].isin(bool_list)]
```

```python
runs = final.groupby('batsman')['batsman_runs'].sum()
```

```python
balls = final.groupby('batsman')['batsman_runs'].count()
```

```python
strike = (runs/balls)*100
```
