```python
import pandas as pd
```

```python
df = pd.read_csv('tennis.csv')
```

```python
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
outlook	= LabelEncoder() #labelencoder thi e string ne number ma covert kre automatically basis on that
humidity = LabelEncoder()
windy	= LabelEncoder()
play = LabelEncoder()
```

```python
#now the data of string is converted into the numeric so we will replace the data of it
df['outlook'] = outlook.fit_transform(df['outlook'])
df['humidity'] = humidity.fit_transform(df['humidity'])
df['windy'] = windy.fit_transform(df['windy'])
df['play'] = play.fit_transform(df['play'])
```

```python
df
```

```python
feature = ['outlook','humidity','windy']
```

```python
x = df[feature]
```

```python
y = df.play
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
classifier = DecisionTreeClassifier(criterion = 'entropy')
```

```python
classifier.fit(x_train,y_train)
```

```python
classifier.predict(x_test)
```

```python
x_test
```

```python
classifier.score(x_test,y_test)
```

```python
from sklearn import tree
```

```python
tree.plot_tree(classifier)
```

```python
from sklearn import datasets
```

```python
iris = datasets.load_iris()
```

```python
iris
```

```python
print(iris.target_names)
```

```python
print(iris.feature_names)
```

```python
import pandas as pd
```

```python
data = pd.DataFrame({
    'sepal length': iris.data[:,0],
    'sepal width': iris.data[:,1],
    'petal length': iris.data[:,2],
    'petal width': iris.data[:,3],
    'species': iris.target
})
```

```python
data
```

```python
x = data[['sepal length','sepal width','petal length','petal width']]
```

```python
y = data['species']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
clf = RandomForestClassifier(n_estimators=100,criterion='gini')
```

```python
clf.fit(x_train,y_train)
```

```python
clf.predict(x_test)
```

```python
x_test  #0:setosa 1:versicolor 2: virginica
```

```python
clf.score(x_test,y_test)
```
