3d surface plots most imp for ML

3D Surface Plots use fro the Loss function that we use in the ML Models to check the accuracy of the Ml Models

```python
#so we will take any loss function any for an example to plot the graph
# we need x an y values that it should be more than 100 not any dataset for plotting surface we have to create an equation for the loss function so we need random values not any dataset or we dont have to fecth anything so we are creating x datapoints with linspace somewhat we need 100 values
```

```python
x = np.linspace(-10,10,100) #columns
y = np.linspace(-10,10,100) #rows
x
```

```python
y
```

```python
xx, yy = np.meshgrid(x,y) #it combines the rows and cols meand it converts the 1D in to the 2D coordinates stores in the xx , yy and mesh grid coverts 1d into 2d grid means it converts basis on the rows to the cols and vice-versa same for the cols so it stores in to the xx and yy 2D
```

```python
xx.shape
```

```python
z = (xx)**2 + (yy)**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis') #it helps to plot the surface function

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis') #it helps to plot the surface function

fig.colorbar(p)
```

```python
z = np.sin(xx) + np.log(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis') #it helps to plot the surface function

fig.colorbar(p)
```
