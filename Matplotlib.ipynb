# Color, Hex and Linestyle and width and marker and marker size and Label


```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='yellow')
#we can add color into this and we can add hexadecimal code into this of any color

plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'],color='green', linestyle='dashdot',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='orange', linestyle='dashdot', linewidth=2)
#we have here linestyle which tells about which style of line is this if i want to change the style of line from default than we can use dotted or dashed
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'],color='green', linestyle='solid',linewidth=3,marker='o',markersize='10')
plt.plot(batsman['index'],batsman['RG Sharma'],color='orange', linestyle='dashdot', linewidth=2, marker='D')
#if we want to see the points here we call points as the marker so we will insert marker and write D as diamond shape points and it could be any shape we have written
plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
```

```python
from matplotlib.lines import lineStyles
plt.plot(batsman['index'],batsman['V Kohli'],color='green', linestyle='solid',linewidth=3,marker='o',markersize='10',label='Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='orange', linestyle='dashdot', linewidth=2, marker='D',label='Rohit')
#for plotting label k koi aapdo graph jove to khbr pade koni kai line che so we use label and name aapisu e label ne and if we want to show the label in the graph to identify which line is use for which label so we have to call the function named legend  to show the label in the graph


plt.title('Rohit Sharma vs Virat Kohli Career Comparison')
plt.xlabel('Season')
plt.ylabel('Runs Scored')


plt.legend(loc='best')
#this means this function shows the labels in the graph which line show which info and it have default parameter loc = best means it by default take the position where it have empty space it would take that and we can move the location also according to it
```

Limiting Axis

```python
price = [48000,54000,57000, 49000, 47000, 45000, 4700000]
#aatla moto no hoi tyare 47 lakh to ena same aa badha no's to nana laagse ne as variation bou vadhare che so small changes ne ek simlar j line ma plot karse as bou vadhare difference e value ma and nani values che ema etlo difference nathi so e straight line ma j batavse and pelo difference dekhase if mare evo variation nathi jovo to aapde graph ne trim karisu if aapde small variations jova hoi to

year =  [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,75000)
plt.xlim(2017,2019)

#if mare pela je variations notu aavtu and variations jovu hoi small values ma to hu limit lagavis so mne variations jova malse and aapde bane axis ni limit lagai sakisu

#generally aapde tyare krie jyare koi outlier value aavi jai and variations show na thai tyare use krie che
```

Seaborn graph tells about how much the bill and tip is and the it tells how much no of people are there according to the size

```python
#importing seaborn library

tips = sns.load_dataset('tips')

# scatter plot is slower than plot

plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*30)
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

#this tells about the linear relation that bill increases tip also increases and particular dots represent the persons and if i want to represent a size how many person comes
```

```python
#scatter plot using plt.plot
#faster for higher dataset

plt.plot(tips['total_bill'],tips['tip'],'o')

#if we dont add 'o' it will show the lines but if we add this it will add o so it will look same like scattter plot
```

```python
#color and label

df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2 ,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2)

#we have calculated some maths and change the positions on x axis so we can show multiple bar graphs along side with it 2015 is 0.2 postion before and 2016 same position and 2017 is after 0.2 postion and the we have changed the values of x axis to the 0 to 4 as its shape first index is 5 so we can get easily to know about it and easy to get visual representation

#so now we willreplcae the no's with the batsman names so we will use the xticks

plt.xticks(np.arange(df.shape[0]),df['batsman'])

#xticks ma pela je value hoi e and replace karvani hoi e add karvani

#if we want to show the o/p not the extra text so we will use show

plt.show()
```

```python
#a problem

child = [10,20,40,10,30]
colors=['red red red red red red','blue blue blue blue','green green green green','yellow yellow yellow ','pink']

plt.bar(colors,child,color='orange')

plt.xticks(rotation = 'vertical' )

#aana thi x axis na badha naam vertical dekhai dese and e naam che ne bou space nai le width na format ma and readability enhance thai jase
```

```python
#stacked bar chart

plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=df['2017']+df['2016'],label='2015')

plt.legend()

plt.show()

#je bottom k saav niche rehvanao hoi e sauthi niche lakhvanu and label aapde aapisu kyo color su indicate k label kre che and aapde bottom ma je sauthi niche vadu hase e lakhisu and 2015 je sauthi upar che enu bottom 2016 and 2017 bane che so addition thase bane nu
```

# Pie Chart
1. Univariate/Bivariate
2. Categorical vs Numerical
3. Use Case - To find contribution on a standard scale

```python
#simple data

data = [23,45,100,20,49]
subs= ['eng','science','maths','sst','hindi']

plt.pie(data,labels=subs)

#evu nathi k badha no sum 100 thavo joie but jetlo sum thai ema e 23 na percent jose
```

```python
df = pd.read_csv('/content/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','orange'])
plt.show()
```

```python
#explode shadow

plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.1,0,0,0,0,0],shadow=True)
plt.show()
```
