### 1. Relational Plot

- to see the statistical relation between 2 or more variables.
- Bivariate Analysis

Plots under this section

- scatterplot
- lineplot

```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px  #for importing dataset we use plotly
```

```python
tips = sns.load_dataset('tips')  #it's a seaborn dataset
```

```python
tips
```

```python
#scatter plot is an axes level function and its for the graph we can plot
#here it will directly provide the labels of the x and y

sns.scatterplot(x='total_bill', y='tip', data=tips,hue='sex',style='time',size='size')
```

```python
#relplot is called relational plot and it will provide the figure level function and it will come in the square shape
#seaborn recommend to use figure level function ne aapde axes level function ma replecate kari sakie

sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
#line plot is basically shows the countinuity in the graph and it will join the points and if we have time in the x-axis than we will always use line plot

gap = px.data.gapminder()
df = gap[gap['country']== 'India']
```

```python
df
```

```python
#axes level function

sns.lineplot(data=df,x='year',y='lifeExp')
```

```python
#using relplot

sns.relplot(data=df,x='year',y='lifeExp',kind='line')
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram
tips
```

```python
# plotting univariate histogram
sns.histplot(data=tips, x='total_bill')
```

```python
#figure level

sns.displot(data=tips, x='total_bill',kind='hist')
```

```python
#bins parameter

sns.displot(data=tips, x='total_bill',kind='hist',bins=2)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram.
# Discrete bins are automatically set for categorical variables

# countplot - it tells count of the categorical column aof the categorical columnnd the histogram

sns.displot(data=tips, x='day',kind='hist')
```

```python
#hue parameter
#hue will show different categories of sex in the histogram as shows by diff colors it will show diff categories to the data as every tip represent diff person

#hue is used to show different categories as an diff colors in the graph and we have comparsion of categories in the histogram

sns.displot(data=tips, x='tip',kind='hist',hue='sex')
```

```python
#element -> step

sns.displot(data=tips, x='tip',kind='hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset('titanic')
```

```python
sns.displot(data=titanic, x='age',kind='hist',hue='sex',element='step')
```

```python
# faceting using col and row -> not work on histplot function as it is axis level
sns.displot(data=tips, x='tip', kind='hist',col='sex',element='step')
```

```python
# kdeplot - kernel density estimation
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate

sns.kdeplot(data=tips,x='total_bill')

#y represents the density of the probability
```

```python
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
# hue -> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

```python
# Rugplot

# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot
# and then shows the count of observations within each rectangle with the fill color

#sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian

sns.kdeplot(data=tips, x='total_bill', y='tip')
```

### 2. Matrix Plot

- Heatmap
- Clustermap

```python
# Heatmap

# Plot rectangular data as a color-encoded matrix

#this dataset is a long dataset and we have to convert the long dataset in to the wide dataset so we have to use pivot table for gaining specific information in rows and the columns and the values we want

temp_df = gap.pivot(index='country',columns='year',values='lifeExp')

#heatmap is axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
# annot

#this will give the Europe country data and there lifeexp and all years and we have convert into the small and wide table
temp_df = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')

plt.figure(figsize=(15,15))
sns.heatmap(temp_df,annot=True,linewidth=0.5, cmap='summer')
#annot true means it will show the values of the lifeexp of per country and there yr with respect to it and linewidth means the gap btween the lines is 0.5 and the cmap will show the color of the heatmap
```

```python
 #Clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.

# This function requires scipy to be available.

iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])

#clustermap will separate the similar and disimilar data it will grp similar data and same with disimilar data
#aama je rows and col ma clustermap karvanu hoi e mention karvanu

#aama rows bhi diff order ma che je similar che values thi e najik che and je disimlar che e values thi dur che

#this clustering is called Dendograms
```
