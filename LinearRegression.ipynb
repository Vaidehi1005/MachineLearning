```python
import pandas as pd
```

```python
from sklearn import linear_model
```

```python
df = pd.read_csv("Term_Insurance.csv")
```

```python
df
```

```python
h = df.Height.mean()
```

```python
df.Height = df.Height.fillna(h)
```

```python
df
```

```python
reg = linear_model.LinearRegression()
```

```python
reg.fit(df[['Age','Height','Weight']],df['Premium'])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.predict([[27,165.56,60]])
```
